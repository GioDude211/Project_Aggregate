Title: Scenario Visualizer

Purpose: Creates a bar chart to visualize different scenarios to compare performance metrics. In this case Revenue vs. Total Expenses to give insights as to which scenario is possible.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
import re

class barchartvisualizer:
    def __init__(self, output_dir="Bar_Charts_Output"):
        self.output_dir = output_dir
        if not os.path.exists(self.output_dir):
            os.makedirs(self.output_dir)

    def _sanitize_currency(self, series):
        """
        Cleans currency strings (e.g., '$1,200.00') into floats.
        If the data is already numeric, it passes through.
        """
        if series.dtype == 'object':
            # Remove $, commas, and whitespace
            return series.replace(r'[\$,]', '', regex=True).astype(float)
        return series

    def generate_chart(self, csv_path, filename="Current_Potential_Metrics_Bar_Chart.png"):
        try:
            df = pd.read_csv(csv_path)
            
            if df.shape[1] < 4:
                raise ValueError("CSV requires 4 columns: Title, Revenue, Expenses, Profit.")

            # 1. Positional Extraction
            scenarios = df.iloc[:, 0]
            
            # 2. Sanitize the Metrics (The Fix)
            # Column 1: Revenue, Column 2: Expenses, Column 3: Profit
            revenue  = self._sanitize_currency(df.iloc[:, 1])
            expenses = self._sanitize_currency(df.iloc[:, 2])
            profit   = self._sanitize_currency(df.iloc[:, 3])
            
            headers = df.columns

            # 3. Setup Logic
            x = np.arange(len(scenarios))
            width = 0.25
            fig, ax = plt.subplots(figsize=(12, 7))

            # 4. Rendering
            # Order: Expenses(Red), Revenue(Green), Profit(Blue)
            ax.bar(x - width, expenses, width, label=headers[2], color='#e74c3c')
            ax.bar(x,         revenue,  width, label=headers[1], color='#2ecc71')
            ax.bar(x + width, profit,   width, label=headers[3], color='#3498db')

            # 5. Aesthetic Scaling (Automatic Y-Axis Formatting)
            ax.set_title('Financial Performance: Strategic Scenarios', fontsize=14, fontweight='bold')
            
            # Format Y-axis to show actual currency markers
            from matplotlib.ticker import StrMethodFormatter
            ax.yaxis.set_major_formatter(StrMethodFormatter('${x:,.0f}'))
            
            ax.set_xticks(x)
            ax.set_xticklabels(scenarios, rotation=15)
            ax.legend(frameon=False)
            ax.grid(axis='y', linestyle=':', alpha=0.6)

            # 6. Deployment
            save_path = os.path.join(self.output_dir, filename)
            plt.tight_layout()
            plt.savefig(save_path)
            plt.close()
            
            return f"Architecture Repaired: {save_path}"

        except Exception as e:
            return f"System Overheat: {str(e)}"

# Execute
architect = barchartvisualizer()
print(architect.generate_chart('Data_Input/Current_Potential_Metrics.csv'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

class AdvancedBarChartVisualizer:
    def __init__(self, output_dir="Bar_Charts_Output"):
        self.output_dir = output_dir
        if not os.path.exists(self.output_dir):
            os.makedirs(self.output_dir)

    def _sanitize(self, series):
        """Removes 'mud' from financial strings and ensures numeric float."""
        if series.dtype == 'object':
            return series.replace(r'[\$,\s]', '', regex=True).astype(float)
        return series.astype(float)

    def generate_topology(self, csv_path, filename="Advanced_BarChart_Current_Potential_Metrics.png"):
        try:
            df = pd.read_csv(csv_path)
            
            # Positional Mapping
            scenarios = df.iloc[:, 0]
            revenue   = self._sanitize(df.iloc[:, 1])
            expenses  = self._sanitize(df.iloc[:, 2])
            net_profit = self._sanitize(df.iloc[:, 3])
            
            headers = df.columns
            x = np.arange(len(scenarios))
            
            fig, ax = plt.subplots(figsize=(14, 8))

            # 1. The Bars: Revenue (Up) and Expenses (Down)
            # We explicitly multiply expenses by -1 to ensure they drop below the 0 line
            ax.bar(x, revenue, width=0.6, label=headers[1], color='#2ecc71', alpha=0.8)
            ax.bar(x, -expenses, width=0.6, label=headers[2], color='#e74c3c', alpha=0.8)

            # 2. The Profit Line: Cutting through the bars
            ax.plot(x, net_profit, color='#3498db', marker='o', linewidth=3, 
                    markersize=8, label=headers[3], zorder=5)

            # 3. Reference Architecture (Zero Line)
            ax.axhline(0, color='black', linewidth=1.2, alpha=0.7)

            # 4. Refinement & Scaling
            ax.set_title('Scenario Analysis: Revenue vs. Outflow Topology', fontsize=16, fontweight='bold')
            ax.set_ylabel('Capital Flow ($)')
            
            # Format Y-Axis for currency
            from matplotlib.ticker import StrMethodFormatter
            ax.yaxis.set_major_formatter(StrMethodFormatter('${x:,.0f}'))
            
            ax.set_xticks(x)
            ax.set_xticklabels(scenarios, rotation=10)
            
            # Legend placement outside the 'mud'
            ax.legend(loc='upper left', bbox_to_anchor=(1, 1), frameon=False)
            ax.grid(axis='y', linestyle=':', alpha=0.5)

            # Final Deployment
            plt.tight_layout()
            save_path = os.path.join(self.output_dir, filename)
            plt.savefig(save_path)
            plt.close()
            
            return f"Strategic Visual Deployed: {save_path}"

        except Exception as e:
            return f"Architectural Failure: {str(e)}"

# --- Feature Execution ---
viz = AdvancedBarChartVisualizer()
# Assuming your CSV structure: [Name, Revenue, Expenses, Net]
print(viz.generate_topology('Data_Input/Current_Potential_Metrics.csv'))
```
